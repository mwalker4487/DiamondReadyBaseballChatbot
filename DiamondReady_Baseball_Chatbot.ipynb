```python
# DiamondReady Baseball Chatbot
# A traditional rule-based chatbot for baseball game preparation

def respond(message):
    message = message.lower().strip()

    # Show what the chatbot can do
    if message in ["help", "what can you do?", "what can you do", "options"]:
        return (
            "I can help with pregame routines, warm-ups, mental preparation, "
            "recovery, and general pregame nutrition. "
            "Try asking me how to warm up."
        )

    # Greetings
    elif any(word in message for word in ["hello", "hi", "hey"]):
        return (
            "Welcome to DiamondReady. I can help you prepare for game day. "
            "Type 'help' to see what I can do."
        )

    # Warm-up questions
    elif any(term in message for term in ["warm up", "warmup", "warm-up"]):
        return (
            "Start with light movement and then progress into dynamic mobility "
            "and baseball-specific movements. Build your intensity gradually "
            "instead of starting at full speed."
        )

    # Nutrition questions
    elif any(term in message for term in ["eat", "food", "nutrition", "meal"]):
        return (
            "Keep pregame nutrition familiar and practical. Choose foods you "
            "already tolerate well, allow enough time to digest, and arrive hydrated."
        )

    # Pregame routine questions
    elif any(term in message for term in
             ["pregame", "pre-game", "before game", "before a game"]):
        return (
            "Keep your pregame routine consistent. Arrive prepared, organize "
            "your equipment, complete your warm-up, review your role, and give "
            "yourself a few quiet minutes to focus before competition."
        )

    # Mental preparation questions
    elif any(term in message for term in
             ["mental", "focus", "nervous", "nerves", "confidence"]):
        return (
            "Keep your mental preparation simple. Focus on the next play instead "
            "of the final outcome. A consistent breathing or focus routine can "
            "also help when you feel rushed or nervous."
        )

    # Recovery questions
    elif any(term in message for term in
             ["recover", "recovery", "after game", "after a game"]):
        return (
            "After competition, cool down, rehydrate, eat appropriately, and "
            "prioritize sleep. The goal is to prepare your body for the next "
            "training session or game."
        )

    # End the conversation
    elif message in ["quit", "exit", "bye", "goodbye"]:
        return "Good luck. Stay prepared and compete one pitch at a time."

    # Handle malformed input
    elif len(message) < 2 or not any(character.isalpha() for character in message):
        return (
            "I couldn't understand that input. Please enter a question using "
            "words, or type 'help' to see what I can do."
        )

    # Handle questions outside the programmed topics
    else:
        return (
            "I don't have a programmed response for that question yet. "
            "Type 'help' to see the topics I currently support."
        )


print("DiamondReady")
print("Traditional Baseball Preparation Chatbot")
print("Type 'help' to see available topics or 'quit' to stop.")
print()

while True:
    playerMessage = input("Player: ")
    answer = respond(playerMessage)

    print("DiamondReady:", answer)
    print()

    if playerMessage.lower().strip() in ["quit", "exit", "bye", "goodbye"]:
        break
```
